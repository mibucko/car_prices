In this file, we present the evaluation metrics for 12 different combinations: 3 datasets and 4 machine learning models.

According to the diagrams below, we conclude that, across all evaluated metrics — lowest MAE, MSE, RMSE, and absolute error, and highest R² — the best-performing model is Random Forest trained on Dataset 3.

Dataset 3 is the dataset in which the least amount of data was removed prior to model training. In this dataset, extreme values were not removed, engine volume was not corrected, and only rows containing more than one missing value were deleted.

A general conclusion from this experiment is that keeping more data can result in better model performance, even when some extreme, or inconsistent values,  or cells with missing values, are present.

```python
import plotly.express as px
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/mibucko/car_prices/main/data/model_evaluation_results.csv"
df = pd.read_csv(url)
df["dataset_model"] = ("D" + df["dataset"].astype(str) + " - " + df["model"])
```

```python
for metric in ["MAE", "MSE", "RMSE", "R2", "largest_error_1"]:
    fig = px.bar(df, x="dataset_model", y=metric,
        hover_data={"dataset_model": True, "dataset": True, "model": True, metric: ":.2f"},
        title=metric)

    fig.update_layout(xaxis_title="Dataset + Model", yaxis_title=metric, xaxis_tickangle=-45)

    fig.show()
```
